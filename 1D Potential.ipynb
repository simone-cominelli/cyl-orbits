```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import lagrange
from scipy.special import jv
```

# Effective Potential

https://linkinghub.elsevier.com/retrieve/pii/S0375960118309587


$$
V_{eff}(r; p_\theta, p_z) = 0.5( p_\theta/r - 0.5B_zr )^2 + 0.5(p_z - A_z(r,Bz))^2
$$

where the vector potential is
$$
A_z(r) = m\omega_C F_0 J_2(\lambda r)\\
A_{\theta} = B_z
$$

in the unit of the code $m=1,\; e=1$. Thus $\omega_C = B_z$

```python
def V_eff(r, pth, pz, Bz):
# parameters: par = [p_theta, p_z, omega, m, H, z_0, theta_0]    
    V = 0.5 *( pth/r - 0.5*Bz*r )*( pth/r - 0.5*Bz*r ) + 0.5*(pz - A_z(r,Bz)) *(pz - A_z(r,Bz))
    return V

def A_z(r,Bz):
    F_0 = 1
    lambd =7.5
    result = Bz*F_0*jv(2, lambd*r)
    return result
```

```python
# parameters: par = [p_theta, p_z, omega, m, H, z_0, theta_0]
pth = 0.3
pz = 0.6
Bz = 4
```

## Effective Potential dependences from $p_\theta$ and $p_z$
1. $p_\theta$ increase the steepning of the initial central repulsive part: $\propto (p_\theta/r - \omega r)^2$
2. $p_z$ increase the height of the double well (<i>due to the bessel function $J_2$</i>)

```python
p_theta = np.arange(0.2, 2.1, 0.4)
p_z = np.arange(0.2, 4.1, 0.4)

dx = 0.001     # resolution
x = np.arange(0.1, 1, dx)

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

for j, pthi in enumerate(p_theta):
    y = V_eff(x, pthi, pz, Bz)
    ax[0].plot(x,y, label= rf'$p_{{\theta}}={pthi:.1f}$' )

ax[0].legend()
ax[0].set_ylim(0,10)
ax[0].set_title(rf'Effective Potential vs $p_\theta$')
ax[0].set_xlabel('Radial distance')
ax[0].set_ylabel('Energy')

for j, pzi in enumerate(p_z):
    y = V_eff(x, pth, pzi, Bz)
    ax[1].plot(x,y, label= rf'$p_{{z}}={pzi:.1f}$' )

ax[1].legend()
ax[1].set_ylim(0,10)
ax[1].set_title(rf'Effective Potential vs $p_z$')
ax[1].set_xlabel('Radial distance')
ax[1].set_ylabel('Energy')
```
