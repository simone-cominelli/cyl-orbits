### trapezoidal method
1. compute the integrand function on equispaced $t_i$.
2. use the standard simpson integration


```python
def simpson(y, dx):
    integral = dx/3*(y[0] + y[len(y)-1])
    for i in range(1, int((len(y)-1)/2)):
        integral = integral + 4*dx/3* y[2*i - 1]
    for i in range(1, int((len(y)-1)/2-1)):
        integral = integral + 2*dx/3*y[2*i]
    return integral

def indefinite_trapezoidal_unequal(f, x):
    """Calculate the integral from the Trapezoidal Rule.

    Args:
        f (function): the equation f(x) sampled for x_i
        n (int): number of intervals of the sample x_i

    Returns:
        result (float): numerical approximation of the definite integral.
    """
    trapez =  0.5*(f[1]+f[0])*(x[1]-x[0])
    r = [ trapez ]
    for i in range(1,len(x)-1):
        trapez = 0.5*(f[1:] + f[:-1])*(x[1:]-x[:-1])
        r.append(r[-1] + trapez)
    return r
```
